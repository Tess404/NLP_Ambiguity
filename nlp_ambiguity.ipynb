```python
import unsloth
from transformers import AutoTokenizer, AutoModelForCausalLM, BitsAndBytesConfig
from unsloth import FastLanguageModel
import torch

device = "cuda" if torch.cuda.is_available() else "cpu"
from huggingface_hub import login
login(token="...")

# LLaMA 
llama_model, llama_tokenizer = FastLanguageModel.from_pretrained(
    model_name = "unsloth/llama-3-8b-Instruct-bnb-4bit",
    max_seq_length = 1024,
    load_in_4bit = True,
)

# BLOOM
bloom_model = AutoModelForCausalLM.from_pretrained("bigscience/bloom-3b", torch_dtype=torch.float16).to(device)
bloom_tokenizer = AutoTokenizer.from_pretrained("bigscience/bloom-3b")

# Mistral
mistral_model_name = "mistralai/Mistral-7B-Instruct-v0.1"

bnb_config = BitsAndBytesConfig(
    load_in_4bit=True,
    bnb_4bit_quant_type="nf4",
    bnb_4bit_use_double_quant=True,
    llm_int8_threshold=6.0,
    llm_int8_enable_fp32_cpu_offload=True,
)

mistral_tokenizer = AutoTokenizer.from_pretrained(mistral_model_name)
mistral_model = AutoModelForCausalLM.from_pretrained(
    mistral_model_name,
    quantization_config=bnb_config,
    device_map="auto",
)

models = {
    "llama": llama_model,
    "bloom": bloom_model,
    "mistral": mistral_model
}

tokenizers = {
    "llama": llama_tokenizer,
    "bloom": bloom_tokenizer,
    "mistral": mistral_tokenizer
}
```

```python
questions_en = [
"Where is Cambridge?",
"Where is Springfield?",
"Where is Victoria?",
"Where is Oxford?",
"Where is San José?",
"Where is Quebec?",
"Where is Perth?",
"Where is Kingston?",
"Where is Melbourne?",
"Where is Newcastle?",

"What is Apple?",
"What is Amazon?",
"What is Java?",
"What is Python?",
"What is Tesla?",
"What is Mercury?",
"What is Windows?",
"What is Oracle?",
"What is Shell?",
"What is Delta?",

"Who is Jordan?",
"Who is Taylor?",
"Who is Morgan?",
"Who is Alex?",
"Who is Kim?",

"Who is the president?",
"Who is the prime minister?",
"Who is the king?",
"Who is the CEO?",
"Who is the leader of the country?"
]

questions_nl = [
"Waar ligt Cambridge?",
"Waar ligt Springfield?",
"Waar ligt Victoria?",
"Waar ligt Oxford?",
"Waar ligt San José?",
"Waar ligt Quebec?",
"Waar ligt Perth?",
"Waar ligt Kingston?",
"Waar ligt Melbourne?",
"Waar ligt Newcastle?",

"Wat is Apple?",
"Wat is Amazon?",
"Wat is Java?",
"Wat is Python?",
"Wat is Tesla?",
"Wat is Mercury?",
"Wat is Windows?",
"Wat is Oracle?",
"Wat is Shell?",
"Wat is Delta?",

"Wie is Jordan?",
"Wie is Taylor?",
"Wie is Morgan?",
"Wie is Alex?",
"Wie is Kim?",
"Wie is de president?",
"Wie is de premier?",
"Wie is de koning?",
"Wie is de CEO?",
"Wie is de leider van het land?"
]


# Using 3 different types of prompts
prompt_types = {
    "basic": lambda q: f"{q} Answer in one or two sentences.",
    "instruction": lambda q: (
        f"Answer the following question as precisely as possible in one or two sentences.\n"
        f"{q}"
    ),
    "forced": lambda q: (
        f"{q} If multiple interpretations exist, choose the most likely one. "
        f"Answer in one or two sentences."
    )
}
```

```python
import json

llama_answers = []
save_answers_llama = "llama_answers.json"
bloom_answers = []
save_answers_bloom = "bloom_answers.json"
mistral_answers = []
save_answers_mistral = "mistral_answers.json"


def get_model_answer(model, tokenizer, prompt, max_new_tokens=50):
    """
    Get the LLM answer based on the selected prompt type.
    """
    formatted_prompt = f"Question: {prompt}\nAnswer:"
    
    inputs = tokenizer(formatted_prompt, return_tensors="pt")
    inputs = {k: v.to(device) for k, v in inputs.items()}
    
    outputs = model.generate(
        **inputs,
        max_new_tokens=max_new_tokens,
        temperature=0.7,
        do_sample=True,
        pad_token_id=tokenizer.eos_token_id
    )
    
    decoded = tokenizer.decode(outputs[0], skip_special_tokens=True)
    return decoded.split("Answer:")[-1].strip()
    
# Get the English prompt answers
for question in questions_en:
    for prompt_name, prompt_func in prompt_types.items():
        prompt = prompt_func(question)

        llama_answer = get_model_answer(llama_model, llama_tokenizer, prompt)
        llama_answers.append({
            "language": "en",
            "question": question,
            "prompt_type": prompt_name,
            "answer": llama_answer
        })

        bloom_answer = get_model_answer(bloom_model, bloom_tokenizer, prompt)
        bloom_answers.append({
            "language": "en",
            "question": question,
            "prompt_type": prompt_name,
            "answer": bloom_answer
        })

        mistral_answer = get_model_answer(mistral_model, mistral_tokenizer, prompt)
        mistral_answers.append({
            "language": "en",
            "question": question,
            "prompt_type": prompt_name,
            "answer": mistral_answer
        })

# Get the Dutch prompt answers
for question in questions_nl:
    for prompt_name, prompt_func in prompt_types.items():
        prompt = prompt_func(question)

        llama_answer = get_model_answer(llama_model, llama_tokenizer, prompt)
        llama_answers.append({
            "language": "nl",
            "question": question,
            "prompt_type": prompt_name,
            "answer": llama_answer
        })

        bloom_answer = get_model_answer(bloom_model, bloom_tokenizer, prompt)
        bloom_answers.append({
            "language": "nl",
            "question": question,
            "prompt_type": prompt_name,
            "answer": bloom_answer
        })

        mistral_answer = get_model_answer(mistral_model, mistral_tokenizer, prompt)
        mistral_answers.append({
            "language": "nl",
            "question": question,
            "prompt_type": prompt_name,
            "answer": mistral_answer
        })

# Save answers
with open(save_answers_llama, "w", encoding="utf-8") as f:
        json.dump(llama_answers, f, indent=2, ensure_ascii=False)
with open(save_answers_bloom, "w", encoding="utf-8") as f:
        json.dump(bloom_answers, f, indent=2, ensure_ascii=False)
with open(save_answers_mistral, "w", encoding="utf-8") as f:
        json.dump(mistral_answers, f, indent=2, ensure_ascii=False)
```
